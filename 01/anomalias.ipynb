#### Importação das bibliotecas ####

```python
from collections import Counter
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.preprocessing import StandardScaler
from sklearn.ensemble import IsolationForest
from sklearn.metrics import classification_report, confusion_matrix
from sklearn.neighbors import LocalOutlierFactor
```

#### Carregando o dataset ####

```python
df = pd.read_csv("data/creditcard.csv")

df.head()
```

Contagem de frequência das nossas classes 

```python
print(f"Frequência de valor na coluna: {df['Class'].value_counts()}")
print("0 = normal   /   1 = fraude")
```

#### Padronização das features ####

As colunas 'Time' e 'Amount' são as únicas que estão numa escala diferente das demais, por isso vamos reatribuir a elas o valor de cada linha escalonizada com o Standard Scaler.

```python
scaler = StandardScaler()

df[['Time', 'Amount']] = scaler.fit_transform(df[['Time', 'Amount']])
```

Dataset após escalonizar essas duas colunas:

```python
df.head()
```

#### Definindo as features ####

```python
X = df[['Time', 'V1', 'V2', 'V3', 'V4', 'V5', 'V6', 'V7', 'V8', 'V9', 'V10', 'V11', 'V12', 'V13', 'V14', 'V15', 'V16', 'V17', 'V18', 'V19', 'V20', 'V21', 'V22', 'V23',	'V24', 'V25', 'V26', 'V27', 'V28', 'Amount']]

print("Features definidas!")
```

### Processo de detecção de anomalias ####

##### Isolation Forest #####

Na Isolation Forest, ele realiza árvores de decisão aleatórias a fim de separar os dados, e quanto menos cortes forem precisos para um dado ser separado, mais tendência a ser uma anomalia ele tem. 

###### Treinamento do modelo ######

```python
clf = IsolationForest(contamination=0.01, random_state=42) # o 'contamination' serve para a gente indicar para o modelo a porcentagem estimada de anomalias em relação ao conjunto de dados

clf.fit(X) # treinando o modelo
```

```python
predictions = clf.predict(X) # enviando as nossas features para o modelo treinado, e ele retorna um array numpy com '1' e '-1', em que '1' são os dados normais e '-1' são as anomalias
print(f"Predictions: {Counter(predictions)}") # contando quantos dados normais e quantas anomalias houveram na nossa predição
```

281.958 dados normais

2.849 anomalias

###### Avaliando a performance do modelo ######

```python
scores = clf.decision_function(X)
print(f"Anomaly Scores: {scores}")
```

```python
predictions[predictions == 1] = 0
predictions[predictions == -1] = 1

print(f"Predictions: {Counter(predictions)}")
```

###### Matriz de confusão ######

```python
cm = confusion_matrix(df['Class'], predictions)
print(cm)
```

281.755 verdadeiros positivos

2.560 falsos negativos

203 falsos positivos

289 verdadeiros positivos

Deixou passar 203 fraudes e deu 2.560 alarmes falsos, ou seja, não é um bom modelo.

```python
target_names = ['Normal', 'Fraude']

print(classification_report(df['Class'], predictions, target_names=target_names))
```

##### Local Outlier Factor #####

O Local Outlier Factor verifica a densidade local dos pontos, ou seja, quanto menos pontos estão próximos de um, mais tendência ele tem a ser uma anomalia. 

###### Treinamento do modelo ######

```python
lof = LocalOutlierFactor(n_neighbors=20, contamination=0.0017) # o 'n_neighbors' define a quantidade de vizinhos a serem verificados em um local de densidade

X_pred = lof.fit_predict(X) # treinando o modelo
```

```python
lof_predictions = pd.Series(X_pred) # transformando o 'X_pred' em um DataFrame para ter uma melhor manipulação
lof_predictions[lof_predictions == 1] = 0 # '0' dados normais
lof_predictions[lof_predictions == -1] = 1 # '1' anomalias

print(f"Predictions: {Counter(lof_predictions)}")
```

284.322 dados normais

485 anomalias

###### Avaliação da performance do modelo ######

```python
cm_lof = confusion_matrix(df['Class'], lof_predictions)
print(cm_lof)
```

Não acertou nenhuma fraude;

492 dados normais foram tratados como anomalias;

Deu 485 alarmes falsos.

```python
print(classification_report(df['Class'], lof_predictions, target_names=['Normal', 'Fraude']))
```

#### Conclusão ####

Tanto Isolation Forest quanto Local Outlier Factor tiveram dificuldade em detectar as fraudes nesse dataset. Isso sugere que as fraudes desse dataset não são bem "isoladas" nem "de baixa densidade" no espaço de features.

Pra esse tipo de problema, técnicas de aprendizado supervisionado seriam mais adequadas.
