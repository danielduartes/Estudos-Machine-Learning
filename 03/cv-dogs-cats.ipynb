### Importação das bibliotecas ###

```python
import numpy as np
import matplotlib.pyplot as plt

from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Conv2D, MaxPool2D, Flatten, Dense, Dropout
from tensorflow.keras.preprocessing.image import ImageDataGenerator
```

### Definindo a forma que as imagens destinadas a treino e validação serão tratadas ###

Foram criados dois 'ImageDataGenerator' pelo fato de que o treino e a validação possuem objetivos diferentes, então não seria adequado utilizar apenas um para os dois. 

Na validação não é necessário rotacionar a imagem, por exemplo, isso é feito apenas no treinamento do modelo.

```python
datagen_train = ImageDataGenerator( # guarda as regras de como cada imagem será tratada
    rescale=1.0 / 255, # divide os pixels por 255, deixando os valores entre 0 e 1 (normalização)
    rotation_range=20, # gira cada imagem aleatoriamente em até 20 graus para um lado ou para o outro (data augmentation)
    horizontal_flip=True, # espelha a imagem horizontal, de forma aleatória (data augmentation)
)
```

```python
datagen_val = ImageDataGenerator(
    rescale=1.0 / 255 # divide os pixels por 255, deixando os valores entre 0 e 1 (normalização)
)
```

O 'rescale' fica por que é uma normalização, e não uma distorção.

### Criação dos geradores de lotes ###

Foram criados dois pelo fato de serem duas pastas, uma para treino e outra para validação. Logo é necessário um gerador de lotes para cada pasta.

```python
train_generator = datagen_train.flow_from_directory( # aponta para uma pasta e lê as imagens dela em lotes, aplicando a receita do 'datagen' em cada uma
    directory="data/dogs-cats/training_set/", # pastas onde as imagens estão
    target_size=(256,256), # redimensiona todas as imagens para o mesmo tamanho
    color_mode="rgb", # define como as imagens serão lidas em termos de canais de cor (vermelho, verde e azul)
    class_mode="categorical", # os rótulos saem em formato de one-hot-encoding 
    batch_size=32, # entrega 32 imagens por vez, em vez das 8 mil de uma vez
    shuffle=True, # embaralhando a ordem de entrega das imagens
)
```

8.005 imagens destinadas a treinamento divididas em duas classes, cachorros e gatos.

```python
val_generator = datagen_val.flow_from_directory( # aponta para uma pasta e lê as imagens dela em lotes, aplicando a receita do 'datagen' em cada uma
    directory="data/dogs-cats/test_set/", # pastas onde as imagens estão
    target_size=(256,256), # redimensiona todas as imagens para o mesmo tamanho
    color_mode="rgb", # define como as imagens serão lidas em termos de canais de cor (vermelho, verde e azul)
    class_mode="categorical", # os rótulos saem em formato de one-hot-encoding 
    batch_size=32, # entrega 32 imagens por vez, em vez das 8 mil de uma vez
    shuffle=False, # embaralhando a ordem de entrega das imagens
)
```

2.023 imagens para validação divididas em duas classes, cachorros e gatos.

```python
model = Sequential([
    Conv2D(32, kernel_size=(3,3), activation='relu', input_shape=(256,256,3)),
    MaxPool2D(pool_size=(2,2)),
    Conv2D(64, kernel_size=(3,3), activation='relu'),
    MaxPool2D(pool_size=(2,2)),
    Conv2D(128, kernel_size=(3,3), activation='relu'),
    MaxPool2D(pool_size=(2,2)),
    Flatten(),
    Dropout(0.5),
    Dense(128, activation='relu'),
    Dense(2, activation='softmax')
])
```

```python
model.compile(
    optimizer='adam',
    loss='categorical_crossentropy',
    metrics=['accuracy']
)
```

```python
history = model.fit(
    train_generator,
    validation_data=val_generator,
    epochs=10
)
```
