#### Importação das bibliotecas #####

```python
import numpy as np
import matplotlib.pyplot as plt

from tensorflow.keras.datasets import mnist
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Conv2D, MaxPooling2D, Flatten, Dense, Rescaling
from tensorflow.keras.utils import to_categorical
```

Carregamento do dataset e separação dos conjuntos de treino e teste

```python
(X_train, y_train), (X_test, y_test) = mnist.load_data()
```

#### Formato do nosso dataset de imagens ####

São 60.000 imagens e todas são 28x28 pixels.

```python
X_train.shape
```

Existe um 4º item fantasma nesse array, que é a quantidade de canais. Mas por as imagens serem em preto e branco esse número é sempre igual a 1, ficando implícito no .shape.

```python
X_train = np.expand_dims(X_train, axis=3)
X_test = np.expand_dims(X_test, axis=3)
```

```python
print(f'Dimensões do conjunto de treino após adicionar a nova dimensão: {X_train.shape}')
print(f'Dimensões do conjunto de teste após adicionar a nova dimensão: {X_test.shape}')
```

#### Normalização dos pixels ####

Os pixels podem ir de 0 a 255, e para deixar entre 0 e 1, podemos dividir por 255.

```python
X_train_scaled = X_train / 255
X_test_scaled = X_test / 255
```

#### One-Hot-Encoding ####

```python
y_train_cat = to_categorical(y_train, num_classes=10) 
y_test_cat = to_categorical(y_test, num_classes=10) 
```

#### Criação do modelo ####

```python
model = Sequential([ # modelo 'Sequential' para empilhar camadas uma após a outra
    Conv2D(32, kernel_size=(3,3), activation='relu', input_shape=(28,28,1)), # representa quantos filtros a camada vai estar ter, aprendendo em paralelo
    MaxPooling2D(pool_size=(2,2)), # reduzindo o tamanho da imagem processada
    Flatten(), # achata as imagens (matrizes) em listas 
    Dense(64, activation='relu'), # camada de rede neural com 64 neurônios que combina tudo que foi extraído das convoluções
    Dense(10, activation='softmax') # ultima camada com 10 neurônios, um para cada categoria, e o 'softmax' calcula a probabilidade da imagem pertencer a cada uma delas
])  
```

#### Compilando o modelo ####

Na compilação a gente diz para o modelo como a gente quer que ele avalie os erros durante o seu treino e como ajustar os pesos.

```python
model.compile(
    optimizer='adam',
    loss='categorical_crossentropy',
    metrics=['accuracy']
)
```

#### Treinamento do modelo ####

```python
model.fit(X_train_scaled, y_train_cat,
           epochs=5, # quantas vezes o modelo vai passar pelo conjunto de treino inteiro
           batch_size=32, # treinamento em lotes de imagens, nesse caso, 32 imagens por vez
           validation_split=0.1) # separando uma fração das imagens que vai ser utilizada para validação durante o treinamento
```

#### Avaliação da perf

```python
print("Evaluate on the test data")
results = model.evaluate(X_test_scaled, y_test_cat, batch_size=64)
print("test loss, test acc: ", results)
```
